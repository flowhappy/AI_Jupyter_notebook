## 实现线性预测
    任务：给予generated_data.csv的数据，建立线性回归模型，预测x=3.5对应的y值，评估模型的表现

```python
# load data
import pandas as pd
data = pd.read_csv('generated_data.csv')
```

```python
data.head()
```

```python
x = data.loc[:,'x']
y = data.loc[:,'y']

print(x,y)
```

```python
# visualize the data
from matplotlib import pyplot as plt
plt.figure(figsize=(5, 5))
plt.scatter(x,y)
plt.show()
```

```python
# setup linear regression model
import sklearn
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
```

```python
# 如果数据不是二维矩阵，需要转换
import numpy as np
x = np.array(x)
x = x.reshape(-1,1)
y = np.array(y)
y = y.reshape(-1,1)
print(x.shape,y.shape)
```

```python
# 建立模型
lr_model.fit(x,y)
```

```python
# 预测一下数据
y_predict = lr_model.predict(x)
print(y_predict)
```

```python
y_3 = lr_model.predict([[3.5]])
print(y_3)
```

```python
# 打印a,b
print(f'a={lr_model.coef_}')
print(f'b={lr_model.intercept_}')
```

```python
# 评估模型
from sklearn.metrics import mean_squared_error,r2_score
MSE = mean_squared_error(y,y_predict)
R2 = r2_score(y,y_predict)
print(MSE,R2)
```

```python
# 图形展示
plt.figure()
plt.plot(y,y_predict)
plt.show()
```
